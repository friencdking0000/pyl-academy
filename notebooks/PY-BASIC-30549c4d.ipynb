### 寫死的絕對路徑
path 是 CSV 檔的完整路徑，但它寫的是講師電腦上 TSIC 使用者的資料夾。你的電腦沒有這個資料夾，所以下一行開檔就會失敗。不要去建立假的資料夾，正確做法是改成自己的路徑。

### 開檔並讀取全部內容
open 函式用唯讀模式、utf 8 編碼打開檔案，read 把全部內容讀成一個很長的字串，存進 a。

```python
f = open(path,'r',newline='',encoding='utf-8')
```

```python
a = f.read()
```

### 作者的探索筆記
這幾行被註解掉的 print，是作者一步一步確認資料長什麼樣子的過程：先看內容、再看型別、再看長度。拿到陌生資料時，這是很好的習慣。

```python
# print(a) #先查看檔案原始內容是啥
```

### 用換行切成一列一列
split 函式括號反斜線 n，會在每個換行的地方切開，把長字串變成一個列表，每個元素是一列資料。

```python
# print(type(a))# 再查看他類型是什麼
# print(len(a))#看有多長
b = a.split("\n") #決定用\n作為分割的依據
```

### 繼續確認切好的資料
又是一段被註解掉的檢查：印出前幾列和 Excel 對照，再看總共有幾列。確認資料切對了，才往下處理。

### 再用逗號切成一格一格
迴圈處理每一列，用逗號再切一次，得到每一格的資料，放進 c。只有 i 小於 39 時才印出第九欄，也就是人口數女。原本的檔案切完最後會多一個空字串，這個條件剛好避開它，否則會出現 IndexError。

```python
# print(b) #看一下分割完的內容
# print(b[0]) #第一格與excel內容去比對
# print(b[1])
# print(b[2])# 比對完前三筆確認是我要的
# print(len(b)) #查看總體的數量(列數)
#print
# print(b[3])
# print(len(b[3]))
c =[]#把每一列的每一行單獨存成list
for i in range(len(b)):#處理每一行
    c.append(str(b[i]).split(","))#把每行的數值換成str在透過split把,分隔的變成單獨的列表元素
    if( i <39): #過濾掉頭跟尾純中文的
        print(c[i][8])
```

### 取出特定一格
c 中括號 1 中括號 1 是第二列的第二欄，也就是臺南市那一列的里數，印出 649。

```python
print(c[1][1])
```
