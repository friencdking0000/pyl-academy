### 匯入套件與中文字型
os 用來處理路徑，pandas 簡稱 pd 用來讀表格，plt 用來畫圖。這次用 rcParams 設定中文字型，效果和 Day4 的 plt 點 rc 一樣。需要先安裝 pandas 和 openpyxl。

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft JhengHei']
plt.rcParams['axes.unicode_minus'] = False
```

### 讀取 Excel
先找出這支程式所在的資料夾，再接上 data 點 xlsx。read_excel 把 Excel 讀成 DataFrame 表格。header 等於 None 代表不把任何一列當欄名，skiprows 等於 1 跳過第一列標題。

```python
script_dir = os.path.dirname(os.path.abspath(__file__))
excel_path = os.path.join(script_dir, 'data.xlsx')
df = pd.read_excel(excel_path, header=None, skiprows=1)
```

### 用 iloc 取出四個欄位
iloc 用位置取資料，冒號逗號 0 代表所有列的第一欄。astype int 轉成整數，tolist 轉成一般列表。第 0 欄是年、第 1 欄是月、第 8 欄是出生數、第 9 欄是死亡數。

```python
years  = df.iloc[:, 0].astype(int).tolist()
months = df.iloc[:, 1].astype(int).tolist()
births = df.iloc[:, 8].astype(int).tolist()
deaths = df.iloc[:, 9].astype(int).tolist()
```

### 算差值、決定顏色
labels 把年和月組成像 110 斜線 01 的標籤。diff 用 zip 把出生和死亡配對相減。colors 判斷每個差值：大於等於 0 用藍色，小於 0 用紅色。

```python
labels = [f'{y}/{m:02d}' for y, m in zip(years, months)]
diff = [b - d for b, d in zip(births, deaths)]
colors = ['#2E86DE' if v >= 0 else '#E74C3C' for v in diff]
```

### 畫長條圖
x 是每根長條的位置編號。figure 設定圖的寬高，plt 點 bar 畫長條並套用顏色，axhline 在 0 的位置畫一條黑色水平線當基準。

```python
x = range(len(labels))

plt.figure(figsize=(14, 5))
plt.bar(x, diff, color=colors)
plt.axhline(0, color='black', linewidth=0.8)
```

### 刻度與標題
xticks 把月份標籤放在 X 軸，rotation 等於 75 讓文字斜放才不會擠在一起。再加上兩軸名稱和圖表標題。

```python
plt.xticks(x, labels, rotation=75, fontsize=8)
plt.xlabel('年/月')
plt.ylabel('出生 - 死亡 (人)')
plt.title('湖西鄉 每月自然增加人口 (出生 - 死亡)')
```

### 自訂圖例
因為顏色是一根一根指定的，matplotlib 不知道藍和紅代表什麼。所以用 Patch 自己做兩個色塊，告訴讀者藍色是正成長、紅色是負成長。

```python
from matplotlib.patches import Patch
plt.legend(handles=[
    Patch(color='#2E86DE', label='正成長 (出生 > 死亡)'),
    Patch(color='#E74C3C', label='負成長 (出生 < 死亡)'),
])
```

### 排版與顯示
tight_layout 自動調整邊界，避免文字被切掉，最後用 plt 點 show 顯示圖表。

```python
plt.tight_layout()
plt.show()
```
