### 用程式所在資料夾找檔案
原檔的路徑是 C 槽 Users TSIC 開頭的絕對路徑，換電腦就找不到。修正版從 pathlib 匯入 Path，用 Path 括號 __file__ 點 parent 找到這支程式自己所在的資料夾，再接上 CSV 檔名。這樣不管你從哪個資料夾執行，都找得到旁邊的 CSV。

```python
# 修正 1:路徑寫死成講師電腦 C:\Users\TSIC\... → 換台電腦就 FileNotFoundError;改用 Path(__file__).parent 找「這支程式所在資料夾」的 CSV,從哪裡執行都找得到
# 修正 2:open() 之後沒有關檔 → 改用 with open(...),讀完自動關閉
# 修正 3:原註解說 i < 39 是「過濾掉頭跟尾純中文的」,但表頭其實照樣印出;它真正的作用是避開檔案最後切出來的空字串(否則 IndexError)→ 改成直接檢查這一列有沒有第 9 欄
from pathlib import Path
path = Path(__file__).parent / '985ed9fd-c6a3-4421-9886-79621e170322.csv'  # 修正 1
```

### 用 with 開檔
原檔用 open 開檔之後一直沒有 close。修正版改成 with open，區塊結束時 Python 會自動幫你關檔。這是讀檔案最推薦的寫法，Day4 的台南人口統計點 py 也是這樣寫。

```python
with open(path,'r',newline='',encoding='utf-8') as f:  # 修正 2
    a = f.read()
# print(a) #先查看檔案原始內容是啥
# print(type(a))# 再查看他類型是什麼
# print(len(a))#看有多長
b = a.split("\n") #決定用\n作為分割的依據
# print(b) #看一下分割完的內容
# print(b[0]) #第一格與excel內容去比對
# print(b[1])
# print(b[2])# 比對完前三筆確認是我要的
# print(len(b)) #查看總體的數量(列數)
#print
# print(b[3])
# print(len(b[3]))
c =[]#把每一列的每一行單獨存成list
```

### 修正 i 小於 39 的判斷
原檔註解說 i 小於 39 是在過濾掉頭尾的純中文，但其實表頭那列照樣會印出「人口數女」。它真正的作用，是避開檔案最後切出來的空字串，那一列只有一格，取第九欄會 IndexError。修正版直接檢查這一列的長度大於 8，也就是真的有第九欄才印，不用依賴 39 這個神祕數字。驗證方法：從別的資料夾執行，會印出人口數女、38 個數字，最後是 649。

```python
for i in range(len(b)):#處理每一行
    c.append(str(b[i]).split(","))#把每行的數值換成str在透過split把,分隔的變成單獨的列表元素
    if len(c[i]) > 8: # 修正 3:這一列真的有第 9 欄才印;檔案結尾的空行切出來只有 1 格,會被跳過
        print(c[i][8])
print(c[1][1])
```
