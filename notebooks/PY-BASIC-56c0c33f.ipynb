### 匯入工具與中文字型
csv 模組專門讀 CSV 檔，Path 用來組路徑，plt 用來畫圖。後兩行設定中文字型，讓區名不會變成方框。

```python
import csv
from pathlib import Path
import matplotlib.pyplot as plt
plt.rc('font', family='Microsoft JhengHei')
```

### 找到同資料夾的 CSV
Path 括號 __file__ 是這支程式自己的位置，點 parent 是它所在的資料夾，再用斜線接上 CSV 檔名。這樣不管放在誰的電腦，只要 CSV 和程式在一起就讀得到。districts 和 populations 兩個空列表，等一下分別裝區名和人口數。

```python
plt.rcParams['axes.unicode_minus'] = False
# Step 1: Read CSV
csv_path = Path(__file__).parent / '985ed9fd-c6a3-4421-9886-79621e170322.csv'
districts = []
populations = []
```

### 開檔與讀表頭
with open 開啟檔案，離開區塊時會自動關檔。csv 點 reader 幫你把每一列切好。next 先讀出第一列當表頭，strip 去掉檔案開頭看不見的 BOM 字元。

### 逐列整理資料
迴圈一列一列處理，dict 搭配 zip 把表頭和這一列配對，就能用欄名取資料。遇到「臺南市」這一列是全市合計，用 continue 跳過，不然圓餅圖會多出一塊占一半的假資料。其他區把名稱和人口數放進列表，人口要用 int 轉成整數。

### 畫圓餅圖
figure 設定圖的大小為十乘十英寸。plt 點 pie 用人口數畫扇形，labels 放區名，autopct 讓每塊顯示百分比到小數第一位。

```python
with open(csv_path, newline='', encoding='utf-8') as f:
    reader = csv.reader(f)
    header = [col.strip('\ufeff') for col in next(reader)]
    for row in reader:
        data = dict(zip(header, row))
        if data['區域別'] == '臺南市':
            continue
        districts.append(data['區域別'])
        populations.append(int(data['人口數總計']))

# Step 2: Plot pie chart
plt.figure(figsize=(10, 10))
plt.pie(populations, labels=districts, autopct='%1.1f%%')
```

### 標題、比例與顯示
加上標題，axis equal 讓圓餅圖是正圓不會變橢圓，最後 plt 點 show 顯示。

```python
plt.title('臺南市各區人口比例')
plt.axis('equal')
plt.show()
```
